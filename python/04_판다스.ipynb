## 3) 결측치

````python
import pandas as pd
import numpy as np

data = pd.Series([1, np.nan])
print(data)
````

### (1) 결측치 발생

````python
data1 = pd.Series(data=[10,20,30])
data2 = pd.Series(data=[1,2,3,4])

data3 = data1 + data2
print(data3) # 자릿수가 안맞으면 결측으로 표시함
````

예제
* 데이터가 10, 20, 30인 시리즈 생성, 인덱스는 부산, 울산, 대구으로 설정
* 데이터가 1, 2, 3, 4인 시리즈 생성, 인덱스는 광주, 부산, 울산, 대구으로 설정
* 데이터의 덧셈

````python
data1 = pd.Series(data=[10,20,30], index= ['부산','울산','대구'])
data2 = pd.Series(data=[1,2,3,4], index= ['광주','부산','울산','대구'])

data3 = data1 + data2
print(data1)
print(data2)
print(data3)
# 연산은 인덱스 기준으로 이루어진다.
````

````python
data_list = [
    ['김영철','M',24,179.4],
    ['송윤지','F',31, 161.0],
    ['임수현','F',26, 174]
]
data1 = pd.DataFrame(data_list, columns=['이름','성별','나이','키'])
# data.columns = ['이름','성별','나이','키']
data1
````

````python
# 열 딕셔너리 이용
data_col_dict = {
    '이름': ['김영철','송윤지', '임수현'],
    '성별': ['M','F','M'],
    '나이': [24,31,26],
    '키': [179.4, 161.0, 174]
}
data2 = pd.DataFrame(data_col_dict)
data2
````

````python
# 행 딕셔너리 이용
data_row_list = [
    {'이름': '김영철', '성별': 'M','나이':24,'키': 179.4},
    {'이름': '송윤지', '성별': 'F', '나이':31, '키': 161.0},
    {'이름': '임수현', '성별': 'F', '나이': 26, '키': 174}
]
data3 = pd.DataFrame(data_row_list)
data3
````

````python
# 이름 열을 인덱스로 설정하기(덮어쓰기 해야 함)
data4 = data1.set_index('이름')
data4
````
